Постановка задачи:

Написать уравнение плоской фигуры, состоящей из всех точек, произведение расстояний которых   
до двух данных точек $F_1$ и $F_2$ (|$F_1F_2$| = $2b$) есть величина постоянная, равная $a^2$.

Несложными преобразованиями (см. файлик) придем к следующей форме записи:  
$(x + y)^2 - 2b^2x^2 + 2b^2y^2 + b^4 = a^4 \ \ \ (1)$  
где $x$,$y$ - переменные, $b = const$ - задана по условию, $a$ - параметр.

```python
"""
Импортируем библиотеки для отрисовки 
"""

import matplotlib.pyplot as plt
import numpy as np
```

Работаем с плоской фигурой $\Longrightarrow$ будем работать в двумерном векторном   
пространстве $V = \mathbb{R}^2$ над полем вещественных чисел $\mathbb{R}$

```python
"""
Зададим функцию отрисовки
"""


def draw(points, functions=None, implicit_functions=None):
    fig, ax = plt.subplots()

    #xlim и ylim можно редактировать, чтобы поменять параметры размеров плоскости
    xlim, ylim = (-250, 250) , (-250, 250)

    ax.set_xlim(*xlim) #границы по оси Ох
    ax.set_ylim(*ylim) #границы по оси Оу
    ax.set_aspect('equal') #нормирование масштаба по осям

    #отрисовываем точки
    for p in points:
        ax.scatter(*p)
    
    #отрисовываем функции
    if functions:
        x = np.linspace(*xlim, 400)
        for f in functions:
            ax.plot(x, f(x))

    if implicit_functions:
        xs = np.linspace(*xlim, 400) #плотные значения по x
        ys = np.linspace(*ylim, 400) #плотные значения по y
        X, Y = np.meshgrid(xs, ys) #две матрицы плотных значений

        for g in implicit_functions:
            Z = g(X, Y)
            print(f"Z min/max = {float(Z.min())}, {float(Z.max())}")
            ax.contour(X, Y, Z, levels=[0])
        
    plt.show()
```

```python
"""
Задаим плоскость размером 500 x 500, эти параметры можно будет изменить
"""
points = [] #изначально множество точек пустое (очевидно)

draw(points)

```

```python
"""
Зададим две пространственные координаты
"""

x_coordinate, y_coordinate = 40, 150

F1 = (x_coordinate, y_coordinate)

x_coordinate, y_coordinate = -37, 52

F2 = (x_coordinate, y_coordinate)
```

```python
"""
Помещаем координаты на плоскость
"""

points.append(F1)
points.append(F2)

draw(points)
```

```python
"""
Исходя из положения стартовых точек зададим систему координат
"""

x1, y1 = F1
x2, y2 = F2

#нам нужно задать такую функцию, каждое значение которой будет лежать на прямой, соединяющей наши точки, соответственно 

k = (y2 - y1)/(x2 - x1) #вычислили угловой коэффициент
#т.к. y = kx + b => b = y - kx
b = y1 - k * x1 #строим прямую от первой точки
#тогда функция прямой это 
f = lambda x: k*x + b

functions = [f] 

draw(points, functions)
```

```python
"""
Зададим теперь ортогональную прямую, пересекающую первую пряму в точке, соответствующей
расстоянию между двумя точками
"""

x0, y0 = (x1 + x2)/2, (y1 + y2)/2 
k0 = -1 / k #ортогональная прямая к прямой с наклоном k имеет наклон -1/k
b0 = y0 - k0 * x0
f0 = lambda x: b0 + k0 * x
functions.append(f0)

"""
Отметим заодно точку пересечения двух наших прямых
"""

F0 = (x0, y0)

points.append(F0)

draw(points, functions)
```

Отлично! Мы программно показали, что для любых двух точек можно задать удобную, подходящую нам систему координат, выведем теперь явно расстояние от точки F1 до точки F2

```python
from math import sqrt

B2 = sqrt((x2 - x1)**2 + (y2 - y1)**2)
print(f"Расстояние от F1 до F2 = {B2:.2f}")

B = B2/2
print(f"Расстояние от начала координат до любой из точек F1, F2 = {B:.2f}")
```

Осталось только показать, как будет выглядеть плоская фигура, состаящая из всех точек, расстояние которых до точек F1 и F2 - есть постоянная величина, равная $a^2$

То есть нам просто нужно показать график неявной функции $(1)$, то есть такое множество точек $(x,y)$, в котором выполняется равенство $(1)$ 

```python
"""
Вспомним вывод из файлика и выведем ещё раз значение B
"""

print(f"B = {B:.2f}")
```

```python
"""
Зададим значение параметру a
"""

a = B

"""
Задаим функцию перевода в НАШУ новую систему координат
"""
#tlc - to local coordinats
def tlc(X, Y, F1, F2, F0):
    x1, y1 = F1
    x2, y2 = F2
    x0, y0 = F0

    theta = np.arctan2(y2 - y1, x2 - x1) #угол наклона
    si, co = np.sin(theta), np.cos(theta)

    X_center = X - x0
    Y_center = Y - y0

    U = co*X_center + si*Y_center
    V = -si*X_center + co*Y_center

    return U, V


#а теперь зададим нашу неявную функцию, а затем отобразим ее на графике

def g(X, Y):
    U, V = tlc(X, Y, F1, F2, F0)
    return (U**2 + V**2)**2 - 2*(B**2)*(U**2 - V**2) + B**4 - a**4

implicit_functions = [g]

    
draw(points, functions, implicit_functions)
```
